```python
create database NetflixDB;
select * from netflix;

select country, max(duration) as max_duration
from netflix
where country IS NOT NULL
group by country
order by  max_duration desc;

SELECT country,
       MAX(CAST(SUBSTRING_INDEX(duration, ' ', 1) AS UNSIGNED)) AS max_minutes
FROM netflix
WHERE duration LIKE '%min'
GROUP BY country
order by max_minutes desc;


select country,
      MAX(CAST(substring_index(duration, ' ',1) AS UNSIGNED)) as max_duration
from netflix
where duration like "%min"
group by country
order by max_duration desc;

select release_year
from netflix;

select country,type,release_year,title,cast
from netflix
where release_year = 2021
AND country IS NOT NULL
AND country = "India"
order by country,release_year desc;


select type ,count(*)
from netflix
group by type;

select country, count(*) as total_content
from netflix
group by country
order by total_content desc
limit 10;

select  country,cast, count(cast) as total_cast
from netflix
group by country,cast
order by total_cast desc;

select director, count(*) AS total
from netflix
where director IS NOT NULL
group by director
order by total desc
limit 10;


select min(release_year) , max(release_year)
from netflix;

select  rating, count(*) as Total_rating
from netflix
group by rating
order by Total_rating desc
limit 10;

--- Movies with longest duration

select type, title,duration
from netflix
where type = "Movie"
order by duration desc
limit 10;


select type, title, duration 
from netflix
where type = "TV Show"
order by duration desc
limit 10;
select * from netflix;


select 
 sum(case When country = ''  then 1 else 0 end) as misssing_country,
 sum(case When director = ' ' then 1 else 0 end) as missing_director,
 sum(case When cast = '' then 1 else 0 end) as missing_cast
 From netflix;
 
 
 

 
```
